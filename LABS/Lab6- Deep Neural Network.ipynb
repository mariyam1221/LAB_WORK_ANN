# Deep neural network to find the weights and bias for AND Gate

```python
import tensorflow as tf

# Define the inputs and targets for the AND gate
inputs = tf.constant([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=tf.float32)
targets = tf.constant([[0], [0], [0], [1]], dtype=tf.float32)

# Define a simple DNN model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(2,), activation='sigmoid')
])

# Compile the model
model.compile(optimizer='adam', loss='mse')

# Train the model
history = model.fit(inputs, targets, epochs=1000, verbose=1)

# Print the final weights and bias
weights, bias = model.layers[0].get_weights()
print("Final weights:", weights)
print("Final bias:", bias)
```

# Verify the above weights and bias 


```python
import tensorflow as tf

# Define the AND gate function
def and_gate(input1, input2):
    # Define the inputs as TensorFlow tensors
    inputs = tf.constant([[input1, input2]], dtype=tf.float32)
    
    # Define the weights and bias
    weights = tf.constant([[0.6742911], [0.3377598]], dtype=tf.float32)
    bias = tf.constant(-0.8811855, dtype=tf.float32)  # Threshold value
    
    # Calculate the weighted sum and add bias
    weighted_sum = tf.matmul(inputs, weights) + bias
    
    # Apply the step function
    output = tf.where(weighted_sum >= 0, 1.0, 0.0)
    
    return output.numpy()[0][0]

# Testing the AND gate
print("AND gate output:")
print("0 AND 0 =", and_gate(0, 0))
print("0 AND 1 =", and_gate(0, 1))
print("1 AND 0 =", and_gate(1, 0))
print("1 AND 1 =", and_gate(1, 1))
```
