```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ---- Activation Functions Definitions ----
def linear(x):
    return x

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh_bipolar(x):
    # Bipolar sigmoid equivalent to tanh(x/2)
    return (1 - np.exp(-x)) / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, alpha * x)
```

## Task 2: Implement and Test Functions


```python
# ---- Task 1: Plotting Activation Functions ----
x = np.linspace(-10, 10, 400)
funcs = {
    "Linear": (linear(x), "tab:blue"),
    "Sigmoid": (sigmoid(x), "tab:orange"),
    "Tanh (Bipolar Sigmoid)": (tanh_bipolar(x), "tab:green"),
    "ReLU": (relu(x), "tab:red"),
    "Leaky ReLU (alpha=0.1)": (leaky_relu(x), "tab:purple")
}

fig, axes = plt.subplots(2, 3, figsize=(12, 7))
for ax, (name, (y, c)) in zip(axes.ravel()[:5], funcs.items()):
    ax.plot(x, y, color=c, lw=2)
    ax.set_title(name)
    ax.set_xlabel("Input x")
    ax.set_ylabel("f(x)")
    ax.axhline(0, color="k", lw=0.5)
    ax.axvline(0, color="k", lw=0.5)
    ax.grid(True)
```

```python
# Combined plot panel
ax = axes.ravel()[5]
for name, (y, c) in funcs.items():
    ax.plot(x, y, color=c, lw=1.8, label=name.split(" (")[0])
ax.set_ylim(-2, 2)
ax.set_title("All functions together")
ax.set_xlabel("Input x")
ax.grid(True)
ax.legend(fontsize=8)

fig.suptitle("Comparison of Neural Network Activation Functions", fontsize=14)
plt.tight_layout()
plt.show()
```

## Final results and interpretation


```python
# ---- Task 2: Implement and Test Functions ----
test_inputs = np.array([-10, -2, -0.5, 0, 0.5, 2, 10])
print(f"{'x':>6} | {'Linear':>8} {'Sigmoid':>8} {'Tanh':>8} {'ReLU':>8} {'LeakyReLU':>10}")
print("-" * 58)
for v in test_inputs:
    print(f"{v:6.1f} | {linear(v):8.4f} {sigmoid(v):8.4f} {tanh_bipolar(v):8.4f} "
          f"{relu(v):8.4f} {leaky_relu(v):10.4f}")

# Correctness Check
assert np.allclose(tanh_bipolar(test_inputs), np.tanh(test_inputs / 2))
print("\nCheck: bipolar sigmoid(x) == tanh(x/2) for all test inputs -> True")

# Gradients Evaluation
print("\nDerivatives at selected inputs (why vanishing gradients occur):")
print(f"{'x':>6} | {'Sigmoid':>9} {'Tanh':>9} {'ReLU':>6} {'LeakyReLU':>10}")
for v in [-10, -2, 0, 2, 10]:
    s = sigmoid(v)
    t = tanh_bipolar(v)
    print(f"{v:6.1f} | {s*(1-s):9.6f} {0.5*(1-t*t):9.6f} {1.0 if v>0 else 0.0:6.1f} {1.0 if v>0 else 0.1:10.2f}")
```
