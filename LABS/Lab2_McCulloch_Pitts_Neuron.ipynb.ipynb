```python
import numpy as np

def mp_neuron(x, w, theta):
    """McCulloch-Pitts neuron: binary inputs, fixed weights, threshold theta."""
    return 1 if np.dot(w, x) >= theta else 0

inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]

def show(name, fn):
    print(f"{name} gate output:")
    for a, b in inputs:
        print(f"  {a} {name} {b} = {fn(a, b)}")
    print()
```

```python
# ---- Single-Layer Gates ----
AND = lambda a, b: mp_neuron([a, b], [ 1,  1],  2)
OR  = lambda a, b: mp_neuron([a, b], [ 1,  1],  1)
NAND = lambda a, b: mp_neuron([a, b], [-1, -1], -1)
NOR  = lambda a, b: mp_neuron([a, b], [-1, -1],  0)
```

```python
# ---- Two-Layer Gates (Hidden Layer z1, z2 -> Output) ----
def XOR(a, b):
    z1 = mp_neuron([a, b], [ 1, -1], 1)   # a AND (NOT b)
    z2 = mp_neuron([a, b], [-1,  1], 1)   # (NOT a) AND b
    return mp_neuron([z1, z2], [1, 1], 1) # z1 OR z2

def XNOR(a, b):
    z1 = mp_neuron([a, b], [ 1,  1], 2)   # a AND b
    z2 = mp_neuron([a, b], [-1, -1], 0)   # a NOR b
    return mp_neuron([z1, z2], [1, 1], 1) # z1 OR z2

# Display All Outputs
for name, fn in [("AND", AND), ("OR", OR), ("NAND", NAND), ("NOR", NOR), ("XOR", XOR), ("XNOR", XNOR)]:
    show(name, fn)
```

```python
# ---- Automatic Verification against Python logic ----
truth = {
    "AND":  lambda a, b: int(a and b),
    "OR":   lambda a, b: int(a or b),
    "NAND": lambda a, b: int(not (a and b)),
    "NOR":  lambda a, b: int(not (a or b)),
    "XOR":  lambda a, b: int(a != b),
    "XNOR": lambda a, b: int(a == b)
}

fns = {"AND": AND, "OR": OR, "NAND": NAND, "NOR": NOR, "XOR": XOR, "XNOR": XNOR}

print("Verification:")
for k in fns:
    ok = all(fns[k](a, b) == truth[k](a, b) for a, b in inputs)
    print(f"  {k:5s} matches truth table: {ok}")
```
